# Data Preprocessing

```python
# Load Original Data

import pandas as pd
df = pd.read_csv(r'data\CommentDataSet(Media Comments).csv',encoding='cp1252')
```

```python
# Save post-level information

media_type_counts = df['media_type'].value_counts()

media_info = df[['assigned_niche', 'account_handle','timestamp', 'media_type', 'media_product_type', 'like_count', 'views']].drop_duplicates()
media_info.head()

media_info.to_csv(r'data\media_info.csv')
```

```python
# Create comments dataset

import re
comments = df[['assigned_niche', 'account_handle','timestamp','caption','comments']]
comments['comments'] = comments['comments'].apply(lambda x: re.split(r'\n[0-9]+\. ',x[2:]))
comments = comments.explode('comments')
comments['comment_user'] = comments['comments'].apply(lambda x: x.split(': ')[0])
comments['comment_text'] = comments['comments'].apply(lambda x: x.split(': ')[-1])
comments = comments.drop(columns=['comments'])
comments = comments.reset_index()
comments = comments.rename(columns={'index':'media_id'})

comments.to_csv(r'data\comments.csv',index=False)
```
